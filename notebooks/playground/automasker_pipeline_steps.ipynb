# Playground — AutoMasker pipeline steps

Shows each stage used to build the cloth-agnostic mask from a person image:

1. **person** — input photo
2. **DensePose** — body-part map (where the body surface is)
3. **SCHP ATR** — fashion/human parsing (ATR taxonomy)
4. **SCHP LIP** — fashion/human parsing (LIP taxonomy)
5. **agnostic mask** — combined mask for the chosen `mask_type`
6. **overlay** — mask drawn on the person

CatVTON preprocess weights (`DensePose/`, `SCHP/`) download from `zhengchong/CatVTON` on first `load_automasker` call. Runs on CPU (`MASK_DEVICE`).

Display-only: this playground notebook writes nothing to disk.

```python
from fashion_diffusion.config import CATVTON_HF_REPO
from fashion_diffusion.paths import DATA_DIR
from fashion_diffusion.agnostic_mask import (
    MASK_DEVICE,
    load_automasker,
    run_mask_pipeline,
    show_mask_pipeline,
)
```

```python
person_filename = "10336_00.jpg"
split = "train"
mask_type = "upper"  # one of: upper, lower, overall, inner, outer

person_path = DATA_DIR / split / "image" / person_filename

print("==== Input paths:")
print(f"\tperson_path: {person_path}")
print(f"\tCATVTON_HF_REPO: {CATVTON_HF_REPO}")
print(f"\tmask_type: {mask_type}")
print(f"\tMASK_DEVICE: {MASK_DEVICE}")
```

```python
automasker = load_automasker(CATVTON_HF_REPO)
steps = run_mask_pipeline(automasker, person_path, mask_type=mask_type)
show_mask_pipeline(person_path, steps, mask_type=mask_type)
```
